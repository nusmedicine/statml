# Embedding statml widgets in a teaching notebook

Run the cells. There is no kernel communication and no Jupyter extension —
each widget is a static page and this helper builds its URL.

If a widget area looks blank, the cell's output is untrusted. Run the cell
yourself rather than reading saved output.

```python
import sys
sys.path.insert(0, "../python")

import statml_widgets as sw

# Local checkout: run `npm run dev` at the repo root first.
# Once deployed, swap for: sw.use_pages("your-github-user")
sw.use_local()
```

## Week 3 — the sampling distribution of the mean

This starts **empty**, on purpose. Press *Draw one* a few times while you talk
through what a sample mean is — each click draws five observations, collapses
them to their mean, and drops that mean into the pile below. Then press *Play*
to let the remaining samples arrive.

```python
sw.show("clt", dist="exponential", n=5)
```

## The same widget, a different teaching point

One implementation, parameterised per lecture. Zoomed to the mean, `n = 30`:
the histogram and the normal overlay are hard to tell apart.

```python
sw.show("clt", dist="exponential", n=30, view="zoom")
```

## Where convergence is slow

A heavy tail still converges — this Pareto has finite variance — but visibly
more slowly. Worth showing before anyone adopts "$n = 30$ is enough" as a law.

This one is pre-filled with `shown=`: for a figure you are *referring to* rather
than building — a recap, or two end states side by side — ask for the finished
picture instead of making the room wait for it.

```python
sw.show("clt", dist="pareto", n=5, view="zoom", shown=400)
```

## The authoring loop

The intended workflow: open the widget, tune it by hand until it makes the point
you want, press **Copy link**, paste it here. You never hand-write parameters.

```python
sw.show_url("http://localhost:8000/widgets/clt/?dist=bimodal&n=12&view=zoom")
```

## Other helpers

```python
print(sw.url("clt", dist="bimodal", n=30))
display(sw.link("clt", "Open the CLT widget in a new tab", dist="bimodal", n=30))
display(sw.catalogue())
```
